```python
#Criação de uma API com os requisitos solicitados.
#Usaremos o framework FastAPI para construir os endpoints, Pandas para manipulação de dados,
#e SQLAlchemy para gerenciar o armazenamento em banco de dados SQLite (ideal para desenvolvimento e testes).
#Vamos separar o projeto em partes para facilitar o desenvolvimento e os commits.


#### Estrutura do Projeto ####

#    Endpoints da API:

#        Upload de arquivo: Validação do arquivo para evitar duplicidade, processamento dos dados e armazenamento no banco.
#        Histórico de upload: Endpoint para consultar uploads realizados, com filtros por nome do arquivo e data de referência.
#        Buscar conteúdo: Busca paginada com filtros opcionais.

#    Regras de Negócio e Validações:

#        Impedir o upload duplicado com validação por hash.
#        Manipular e armazenar dados em um banco SQLite, garantindo a escalabilidade para um cenário real.

#    Estrutura de Código e Commits:

#        Organizaremos o código em módulos para facilitar a navegação.
#        Cada etapa do projeto será acompanhada por um commit específico, documentando mudanças.


###################
###################
###################
###################


#Passo a Passo do Código
#Primeiro, vamos configurar a estrutura básica e as dependências.

### Passo 1: Instalação de Dependências ###
#Instalando as bibliotecas necessárias:
!pip install fastapi pandas sqlalchemy uvicorn python-multipart

### Passo 2: Estrutura Inicial do Código ####
#Abaixo, vamos configurar o banco de dados, API com FastAPI, e começar a construir o endpoint para o upload de arquivo.
#Configuração do Banco de Dados (SQLite) e Tabelas

#Criando um arquivo database.py para configurar o banco de dados com SQLAlchemy:
# database.py
from sqlalchemy import create_engine, Column, Integer, String, DateTime
from sqlalchemy.ext.declarative import declarative_base
from sqlalchemy.orm import sessionmaker
from datetime import datetime

DATABASE_URL = "sqlite:///./uploads.db"

engine = create_engine(DATABASE_URL, connect_args={"check_same_thread": False})
SessionLocal = sessionmaker(autocommit=False, autoflush=False, bind=engine)
Base = declarative_base()

class FileUpload(Base):
    __tablename__ = "file_uploads"

    id = Column(Integer, primary_key=True, index=True)
    filename = Column(String, unique=True, index=True)
    upload_date = Column(DateTime, default=datetime.utcnow)
    file_hash = Column(String, unique=True, index=True)

class Instrument(Base):
    __tablename__ = "instruments"

    id = Column(Integer, primary_key=True, index=True)
    rpt_date = Column(String)
    ticker_symbol = Column(String)
    market_name = Column(String)
    security_category = Column(String)
    isin = Column(String)
    corp_name = Column(String)

Base.metadata.create_all(bind=engine)

### Passo 3: Configuração do FastAPI ###
#Criando o arquivo principal main.py e configurando o app com FastAPI.

# main.py
from fastapi import FastAPI, UploadFile, File, HTTPException, Depends
from sqlalchemy.orm import Session
###from database import SessionLocal, engine, FileUpload, Instrument
import pandas as pd
import hashlib

app = FastAPI()

# Dependency for session
def get_db():
    db = SessionLocal()
    try:
        yield db
    finally:
        db.close()

#Endpoint de Upload com Validação de Arquivo
#Para verificar se um arquivo já foi carregado, geraremos um hash com hashlib e armazenaremos no banco.
#Esse hash nos permite impedir uploads duplicados.

import io

@app.post("/upload/")
async def upload_file(file: UploadFile = File(...), db: Session = Depends(get_db)):
    if file.content_type not in ["text/csv", "application/vnd.ms-excel"]:
        raise HTTPException(status_code=400, detail="Arquivo deve estar em formato CSV ou Excel")

    # Gera hash para verificar duplicidade
    file_data = await file.read()
    file_hash = hashlib.md5(file_data).hexdigest()
    existing_file = db.query(FileUpload).filter_by(file_hash=file_hash).first()
    if existing_file:
        raise HTTPException(status_code=400, detail="Arquivo já foi enviado")

    # Salva o arquivo no banco
    new_file = FileUpload(filename=file.filename, file_hash=file_hash)
    db.add(new_file)
    db.commit()
    db.refresh(new_file)

    # Carrega dados no DataFrame
    data = pd.read_csv(io.BytesIO(file_data)) if file.content_type == "text/csv" else pd.read_excel(io.BytesIO(file_data))
    # Processa os dados e armazena na tabela Instrument
    for _, row in data.iterrows():
        instrument = Instrument(
            rpt_date=row["RptDt"],
            ticker_symbol=row["TckrSymb"],
            market_name=row["MktNm"],
            security_category=row["SctyCtgyNm"],
            isin=row["ISIN"],
            corp_name=row["CrpnNm"]
        )
        db.add(instrument)
    db.commit()

    return {"message": "Arquivo carregado com sucesso"}

#Endpoint de Histórico de Uploads
#Este endpoint permite buscar uploads anteriores por nome de arquivo ou data de upload.

from typing import Optional
from datetime import datetime

@app.get("/upload/history/")
def upload_history(filename: Optional[str] = None, date: Optional[str] = None, db: Session = Depends(get_db)):
    query = db.query(FileUpload)
    if filename:
        query = query.filter(FileUpload.filename == filename)
    if date:
        query = query.filter(FileUpload.upload_date == datetime.strptime(date, "%Y-%m-%d"))

    history = query.all()
    return [{"filename": file.filename, "upload_date": file.upload_date} for file in history]

#Endpoint para Buscar Conteúdo do Arquivo
#Este endpoint realiza a busca dos dados no arquivo com base nos parâmetros opcionais TckrSymb e RptDt.
#Ele suporta paginação por padrão.

@app.get("/file/content/")
def get_file_content(
    ticker_symbol: Optional[str] = None,
    rpt_date: Optional[str] = None,
    page: int = 1,
    limit: int = 10,
    db: Session = Depends(get_db)
):
    query = db.query(Instrument)
    if ticker_symbol:
        query = query.filter(Instrument.ticker_symbol == ticker_symbol)
    if rpt_date:
        query = query.filter(Instrument.rpt_date == rpt_date)

    results = query.offset((page - 1) * limit).limit(limit).all()
    return [
        {
            "RptDt": item.rpt_date,
            "TckrSymb": item.ticker_symbol,
            "MktNm": item.market_name,
            "SctyCtgyNm": item.security_category,
            "ISIN": item.isin,
            "CrpnNm": item.corp_name,
        }
        for item in results
    ]

#Execução
#Inicie o servidor:

###uvicorn main:app --reload


### Acesse o FastAPI em http://127.0.0.1:8000/docs para testar os endpoints. ###

  ### Essa implementação cobre os requisitos do desafio e é escalável para futuras otimizações. ###
```
